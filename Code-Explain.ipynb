```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()
```

```python
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def ask_llm(prompt):
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "You are a helpful AI teacher."},
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )
    return response.choices[0].message.content

print(ask_llm("Explain AI Agent in 100 words to a 1st grade student."))
```

```python
def calculator(expression):
    try:
        return eval(expression)
    except Exception as e:
        return f"Error: {e}"

def simple_agent(user_input):
    if "calculate" in user_input.lower():
        expression = user_input.lower().replace("calculate", "").strip()
        result = calculator(expression)
        return f"The calculated result is: {result}"
    else:
        return ask_llm(user_input)

# print(simple_agent("calculate 25 * 4"))
print(simple_agent("What is an AI agent explain in 50 words?"))
```

```python
from duckduckgo_search import DDGS

def web_search(query, max_results=3):
    results = []
    with DDGS() as ddgs:
        for r in ddgs.text(query, max_results=max_results):
            results.append({
                "title": r.get("title"),
                "href": r.get("href"),
                "body": r.get("body")
            })
    return results

query = "what are the latest trends in AI startups"
search_results = web_search(query)

for item in search_results:
    print(item["title"])
    print(item["body"])
    print("-" * 50)
```

### Now combine Groq + search tool.

```python
def startup_research_agent(idea):
    search_data = web_search(f"{idea} market trends competitors", max_results=5)

    context = "\n\n".join([
        f"Title: {item['title']}\nSummary: {item['body']}\nURL: {item['href']}"
        for item in search_data
    ])

    prompt = f"""
    You are a startup research analyst.

    Startup idea:
    {idea}

    Web research data:
    {context}

    Give:
    1. Market opportunity
    2. Possible competitors
    3. Target audience
    4. Risks
    5. Final quick opinion
    """

    return ask_llm(prompt)

idea = "AI tutor for Telugu speaking students"
print(startup_research_agent(idea))
```

### Lang Graph

```python
state = {
    "idea": "AI tutor for Hindi speaking students",
    "research": None,
    "risk_analysis": None,
    "final_report": None
}

state["research"] = startup_research_agent(state["idea"])

print(state["research"])
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, END

class StartupState(TypedDict):
    idea: str
    research: str
    risks: str
    final_report: str

def research_node(state: StartupState):
    research = startup_research_agent(state["idea"])
    return {"research": research}

def risk_node(state: StartupState):
    prompt = f"""
    Analyze risks for this startup idea.

    Idea:
    {state['idea']}

    Research:
    {state['research']}

    Give top 5 risks with explanation.
    """
    risks = ask_llm(prompt)
    return {"risks": risks}

def report_node(state: StartupState): 
    prompt = f"""
    Create final startup validation report.

    Idea:
    {state['idea']}

    Research:
    {state['research']}

    Risks:
    {state['risks']}

    Format:
    - Idea Summary
    - Market Opportunity
    - Risks
    - Recommendation  
    - Score out of 10
    """
    final_report = ask_llm(prompt)
    return {"final_report": final_report}

graph = StateGraph(StartupState)

graph.add_node("research", research_node)
graph.add_node("risk_analysis", risk_node)
graph.add_node("final_report", report_node)

graph.set_entry_point("research")
graph.add_edge("research", "risk_analysis")
graph.add_edge("risk_analysis", "final_report")
graph.add_edge("final_report", END)

app = graph.compile()

result = app.invoke({
    "idea": "AI tutor for Telugu speaking students"
})

print(result["final_report"])
```

### Conditional Routing in LangGraph

```python
class StartupState2(TypedDict):
    idea: str
    research: str
    risks: str
    risk_level: str
    improvement: str
    final_report: str

def risk_level_node(state: StartupState2):
    prompt = f"""
    Based on the risks below, classify risk level as only one word:
    LOW, MEDIUM, or HIGH.

    Risks:
    {state['risks']}
    """
    risk_level = ask_llm(prompt).strip().upper()
    return {"risk_level": risk_level}

def improvement_node(state: StartupState2):
    prompt = f"""
    The startup idea has high risk.

    Idea:
    {state['idea']}

    Risks:
    {state['risks']}

    Suggest improvements to reduce risk.
    """
    improvement = ask_llm(prompt)
    return {"improvement": improvement}

def route_by_risk(state: StartupState2):
    if "HIGH" in state["risk_level"]:
        return "improvement"
    return "final_report"

graph2 = StateGraph(StartupState2)

graph2.add_node("research", research_node)
graph2.add_node("risk_analysis", risk_node)
graph2.add_node("risk_level", risk_level_node)
graph2.add_node("improvement", improvement_node)
graph2.add_node("final_report", report_node)

graph2.set_entry_point("research")
graph2.add_edge("research", "risk_analysis")
graph2.add_edge("risk_analysis", "risk_level")

graph2.add_conditional_edges(
    "risk_level",
    route_by_risk,
    {
        "improvement": "improvement",
        "final_report": "final_report"
    }
)

graph2.add_edge("improvement", "final_report")
graph2.add_edge("final_report", END)

app2 = graph2.compile()

result = app2.invoke({
    "idea": "AI tutor for Telugu speaking students"
})

print("Risk Level:", result["risk_level"])
print(result["final_report"])
```

### Human-in-the-loop

```python
def human_approval_node(state):
    print("\n===== HUMAN REVIEW =====")
    print("Idea:", state["idea"])
    print("\nRisks:\n", state["risks"])

    approval = input("\nDo you approve final report generation? yes/no: ")

    if approval.lower() == "yes":
        return {"approval": "approved"}
    else:
        return {"approval": "rejected"}

def rejected_node(state):
    return {
        "final_report": "Human rejected report generation. Please improve the startup idea first."
    }
```

```python
class StartupState3(TypedDict):
    idea: str
    research: str
    risks: str
    approval: str
    final_report: str

def approval_router(state: StartupState3):
    if state["approval"] == "approved":
        return "final_report"
    return "rejected"

graph3 = StateGraph(StartupState3)

graph3.add_node("research", research_node)
graph3.add_node("risk_analysis", risk_node)
graph3.add_node("human_approval", human_approval_node)
graph3.add_node("final_report", report_node)
graph3.add_node("rejected", rejected_node)

graph3.set_entry_point("research")
graph3.add_edge("research", "risk_analysis")
graph3.add_edge("risk_analysis", "human_approval")

graph3.add_conditional_edges(
    "human_approval",
    approval_router,
    {
        "final_report": "final_report",
        "rejected": "rejected"
    }
)

graph3.add_edge("final_report", END)
graph3.add_edge("rejected", END)

app3 = graph3.compile()

result = app3.invoke({
    "idea": "AI tutor for Telugu speaking students"
})

print(result["final_report"])
```
